```python
# Problem Statement
# Using the above data, create a Machine Learning model, which will predict
# if a passenger survives on titanic ship, based on the attributes of that passenger
# like Age, Pclass, Gender, Fare, Emabarkation point, etc.
```

```python
# Output label : Survived
# Input Features : all other features

# Output column is categorical, so this is a Classification Learning Problem
# Output column is measurable, then it will be a Regression Learning Problem
```

```python
df
```

```python
# Step 1
df = pd.read_excel(data)
```

```python
# Step 2
df['Family'] = df['SibSp'] + df['Parch']
```

```python
# Step 3
df = df.drop(columns=['SibSp','Parch','Name','Ticket'])
# Feature Engineering : feature selection or feature extraction
```

```python
# Step 4
df = pd.get_dummies(df,dtype='int',drop_first=True)
```

```python
df.columns
```

```python
# Step 5
X = df[['Pclass', 'Age', 'Fare', 'Family', 'Gender_male','Embarked_Q', 'Embarked_S']]
y = df['Survived']
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=10)
```

```python
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier()
```

```python
clf.fit(X_train,y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
%%writefile app.py
import streamlit as st
import joblib
import pandas as pd

clf = joblib.load('titanic_model')
st.title("Titanic Survival Prediction")

pclass = st.radio("Enter the class on which the passenger traveled",
                  [1,2,3],horizontal= True)
age = st.slider("Enter your age",min_value=1,max_value=90,step=1,value=30)
fare = st.slider("Enter the fare",min_value=5,max_value=150,step=5,value=10)
family = st.s("Number of family members",[0,1,2,3,4,5,6,7,8],horizontal=True)
gender = st.radio("Enter the gender(0:female,1:male)",[0,1],horizontal=True)
embarked = st.radio("Enter the embarkation point",["Q","S","C"],horizontal=True)
if st.button("PREDICT SURVIVAL"):
    passenger = pd.DataFrame({
        'Pclass':[pclass],
        'Age':[age],
        'Fare':[fare],
        'Family':[family],
        'Gender_male':[gender],
        'Embarked_Q': [1 if embarked == 'Q' else 0],
        'Embarked_S' :[1 if embarked == 'S' else 0]})
    result = clf.predict(passenger)
    if result[0] == 1:
        st.success("Passenger Survived!")
    else:
        st.error("Passenger did not Survive!")
```
