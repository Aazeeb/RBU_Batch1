```python
df['Family'] = df['SibSp'] + df['Parch']
```

```python
df = df.drop(columns=['SibSp','Parch'])
```

```python
# Feature Engineering : Feature Extraction
```

```python
df
```

```python
# Based on this data, create a ML model which will predict
# if a person without the provided attributes will likely survive on titanic or not
```

```python
# Input features  : X
# Output label : y
```

```python
X = df[['Pclass','Gender','Age','Fare','Embarked','Family']]
y = df['Survived']
```

```python
X
```

```python
X = pd.get_dummies(X,dtype='int',drop_first=True)
```

```python
X.info()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=7)
```

```python
X_train
```

```python
y_train
```

```python
# based on problem statement, we decide the input features and output labels
# if output label is categorical, it is a classification learning problem
# if output label is measurable, it is a regression learning problem
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
clf = RandomForestClassifier()
```

```python
clf.fit(X_train,y_train)
```

```python
y_pred = clf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
%%writefile app.py
import streamlit as st
import joblib
import pandas as pd

clf = joblib.load('titanic_model')
st.title("Titanic Survival Prediction")

pclass = st.radio("Enter the class on which the passenger traveled",
                  [1,2,3],horizontal= True)
age = st.slider("Enter your age",min_value=1,max_value=90,step=1,value=30)
fare = st.slider("Enter the fare",min_value=5,max_value=150,step=5,value=10)
family = st.radio("Number of family members",[0,1,2,3,4,5,6,7,8],horizontal=True)
gender = st.radio("Enter the gender(0:female,1:male)",[0,1],horizontal=True)
embarked = st.radio("Enter the embarkation point",["Q","S","C"],horizontal=True)
if st.button("PREDICT SURVIVAL"):
    passenger = pd.DataFrame({
        'Pclass':[pclass],
        'Age':[age],
        'Fare':[fare],
        'Family':[family],
        'Gender_male':[gender],
        'Embarked_Q': [1 if embarked == 'Q' else 0],
        'Embarked_S' :[1 if embarked == 'S' else 0]})
    result = clf.predict(passenger)
    if result[0] == 1:
        st.success("Passenger Survived!")
    else:
        st.error("Passenger did not Survive!")
```
